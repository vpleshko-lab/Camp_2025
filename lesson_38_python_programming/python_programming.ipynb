## Part 1: The Foundation: Code Quality & Execution

### 1.1 Python as an Interpreted Language

Python is an **interpreted language**. This means the Python interpreter executes your code line by line, without a separate compilation step. 

* **Compiled Languages** (like C++, Java): The entire source code is translated into machine code by a compiler *before* the program is run. This results in faster execution.
* **Interpreted Languages** (like Python, R): An interpreter reads and executes the source code directly, line by line. This offers more flexibility and makes interactive development (like in Jupyter) possible.

We get the best of both worlds! Python's ease of use is perfect for experimentation. For heavy-duty math, core libraries like **NumPy** and **Pandas** are written in compiled C or Fortran. When you call `numpy.dot()`, you're actually running highly-optimized, pre-compiled C code, not slow Python loops. This is why Python is performant enough for data science.

Because the code is interpreted line by line, a syntax error or a type error in a part of the code that isn't executed won't be raised. This is a key difference from compiled languages, which check the entire program for such errors before running.

````python
def check_value(x):
    if x > 10:
        print("Value is greater than 10.")
    else:
        # This line contains a TypeError, but it won't crash the program
        # unless the 'else' block is actually executed.
        print("Value is not greater than 10, adding a string to a number: " + 5)

# This call will work fine because the 'else' block is never entered.
print("--- Calling with a value that works ---")
check_value(20)

# This call will raise a TypeError because the interpreter now tries to execute the faulty line.
try:
    print("\n--- Calling with a value that will cause an error ---")
    check_value(5)
except TypeError as e:
    print(f"Caught an error as expected: {e}")
````

## Part 2: Managing Your Workspace

### 2.1 Why Use Virtual Environments?

A virtual environment is an isolated Python environment that allows you to manage dependencies for different projects separately.

**Key Concepts:**
- **Dependency Isolation**: Project A can have its own set of packages and versions, completely separate from Project B.
- **Reproducibility**: You can create a file that lists all the specific package versions needed for your project (`environment.yml` for conda, `requirements.txt` for pip). Anyone can then recreate your exact environment, ensuring the code runs the same way everywhere.

Imagine this scenario: 
> Your awesome churn prediction model was built using `scikit-learn==1.0`. You send it to a colleague. Their machine has `scikit-learn==1.4` installed for a different project. When they run your code, it breaks because of a subtle API change between versions. 

Virtual environments solve this problem completely.

### 2.2 Overview of Tooling

- **`conda`**: The recommended tool for data scientists. It's both a package manager and an environment manager. Crucially, it can manage **non-Python dependencies** like `CUDA` (for GPU computing), `MKL` (Intel Math Kernel Library), and other C/Fortran libraries that are essential for high-performance scientific computing.
- **`pyenv`**: Manages different *versions of Python itself* (e.g., Python 3.8 vs 3.9 vs 3.10 on the same machine).
- **`poetry`**: A modern tool for dependency management and packaging in Python. It's excellent for library development but `conda` is often more practical for data science projects due to its handling of non-Python packages.

## Part 3: OOP for Data Scientists

### 3.1 & 3.2: The `DataSet` Class & Magic Methods

Object-Oriented Programming (OOP) allows us to bundle data (attributes) and functions that operate on that data (methods) into objects. For data science, this can be used to create more intuitive and organized wrappers around our data structures.

**Magic Methods** (or *dunder methods*, for double-underscore) are special methods that let you emulate the behavior of built-in types. For example, `__len__` is what allows `len()` to work on your custom object.

Let's build a `DataSet` class that wraps a pandas DataFrame and see how magic methods make it more Pythonic.

````python
import pandas as pd
from typing import List

class DataSet:
    """A wrapper class for a pandas DataFrame to encapsulate data and related operations."""

    # 3.2 Magic Method: The Constructor
    def __init__(self, data: pd.DataFrame):
        """Initializes the DataSet with a pandas DataFrame."""
        if not isinstance(data, pd.DataFrame):
            raise TypeError("Input data must be a pandas DataFrame.")
        self.df = data

    # 3.2 Magic Method: Length
    def __len__(self) -> int:
        """Returns the number of rows in the dataset."""
        return self.df.shape[0]

    # 3.2 Magic Method: Representation
    def __repr__(self) -> str:
        """Provides a developer-friendly string representation of the object."""
        rows, cols = self.df.shape
        return f"<DataSet with {rows} rows and {cols} columns>"
    
    # 3.2 Magic Method: Get Item (for column selection)
    def __getitem__(self, columns: List[str]) -> pd.DataFrame:
        """Allows for dictionary-style column selection."""
        return self.df[columns]
````

#### Demonstrating the Magic Methods

````python
# Create a sample DataFrame
data = {'col_a': [10, 20, 30, 40], 'col_b': ['A', 'B', 'C', 'D'], 'col_c': [1.1, 2.2, 3.3, 4.4]}
sample_df = pd.DataFrame(data)

# 1. Create an instance of our DataSet class
my_dataset = DataSet(sample_df)

# 2. Use the __repr__ magic method (this is called when you print the object)
print("--- Testing __repr__ ---")
print(my_dataset)

# 3. Use the __len__ magic method (this is called by the built-in len() function)
print("\n--- Testing __len__ ---")
print(f"The number of rows is: {len(my_dataset)}")

# 4. Use the __getitem__ magic method (this allows for slicing)
print("\n--- Testing __getitem__ ---")
selected_cols = my_dataset[['col_a', 'col_c']]
print(selected_cols)
````

### 3.3 Method Types: Factories and Helpers

Different types of methods serve different purposes within a class.

- **Instance Method**: The 'normal' method. It operates on an instance of the class (`self`) and can access instance-specific data.
- **Class Method (`@classmethod`)**: Operates on the class itself, not an instance. It receives the class (`cls`) as the first argument. Often used as *factory methods* to create instances of the class in alternative ways.
- **Static Method (`@staticmethod`)**: A function that lives inside the class's namespace but doesn't operate on the instance or the class. It's essentially a regular function that is logically related to the class. Often used for utility or helper functions.

````python
import os

class DataSet:
    """A wrapper class for a pandas DataFrame, now with more methods."""

    def __init__(self, data: pd.DataFrame):
        if not isinstance(data, pd.DataFrame):
            raise TypeError("Input data must be a pandas DataFrame.")
        self.df = data

    def __len__(self) -> int:
        return self.df.shape[0]
    def __repr__(self) -> str:
        rows, cols = self.df.shape
        return f"<DataSet with {rows} rows and {cols} columns>"

    def __getitem__(self, columns: List[str]) -> pd.DataFrame:
        return self.df[columns]
    
    # 1. Instance Method
    def get_summary(self) -> pd.DataFrame:
        """Returns the descriptive statistics of the dataset."""
        return self.df.describe()
    
    # 2. Class Method (Factory)
    @classmethod
    def from_csv(cls, file_path: str):
        """Creates a DataSet instance by loading data from a CSV file."""
        if not cls._is_valid_filepath(file_path, extension='.csv'):
            raise FileNotFoundError(f"File not found or is not a CSV: {file_path}")
        data = pd.read_csv(file_path)
        return cls(data) # Creates an instance of the class
    
    # 3. Static Method (Helper/Utility)
    @staticmethod
    def _is_valid_filepath(path: str, extension: str) -> bool:
        """Checks if a file path exists and has the correct extension."""
        return os.path.exists(path) and path.endswith(extension)
````

#### Demonstrating the Method Types

````python
# Create a dummy CSV file for the example
dummy_data = {'feature1': [1, 2, 3], 'feature2': [4, 5, 6]}
dummy_df = pd.DataFrame(dummy_data)
dummy_csv_path = 'sample_data.csv'
dummy_df.to_csv(dummy_csv_path, index=False)

# 1. Use the @classmethod factory to create an instance
print("--- Testing @classmethod ---")
dataset_from_file = DataSet.from_csv(dummy_csv_path)
print(f"Created instance from file: {dataset_from_file}")
print(dataset_from_file.df.head())

# 2. Use the instance method on the new object
print("\n--- Testing instance method ---")
summary = dataset_from_file.get_summary()
print("Descriptive summary:")
print(summary)

# 3. Use the @staticmethod directly from the class
print("\n--- Testing @staticmethod ---")
print(f"Is '{dummy_csv_path}' a valid path? {DataSet._is_valid_filepath(dummy_csv_path, '.csv')}")
print(f"Is 'nonexistent.txt' a valid path? {DataSet._is_valid_filepath('nonexistent.txt', '.csv')}")
````

### 3.4 Base classes and Inheritance

Using a Base Class for ML Models
In machine learning, different models (like Linear Regression or a Random Forest) share a common lifecycle: you train them with data, and then you use them to make predictions. We can enforce this common structure using a base class.

An Abstract Base Class (ABC) defines a common interface (a set of methods) that all subclasses must implement. This acts as a "contract," ensuring that every model you create will have a consistent API.

1. Defining the Abstract Base Class
We'll use the abc module to create a BaseModel. It will require any inheriting class to have a train and a predict method.

````python
from abc import ABC, abstractmethod
import numpy as np
import pandas as pd

# Define the "contract" for all our models
class BaseModel(ABC):
    """
    An abstract base class that defines the standard interface for our models.
    """
    def __init__(self):
        # The actual trained model object will be stored here
        self.model = None
        print(f"Initialized a {self.__class__.__name__} instance.")

    @abstractmethod
    def train(self, X_train: pd.DataFrame, y_train: pd.Series):
        """
        Trains the model on the provided data.
        This method MUST be implemented by any subclass.
        """
        pass

    @abstractmethod
    def predict(self, X_test: pd.DataFrame) -> np.ndarray:
        """
        Makes predictions on new data.
        This method MUST be implemented by any subclass.
        """
        pass

    def get_info(self):
        """
        A regular method that all subclasses will inherit and can use directly.
        """
        return f"This is a model of type: {self.__class__.__name__}"

# Note: If you try to create an instance of the base class directly, you will get an error.
# This is because it has abstract methods that are not implemented.
try:
    base_model = BaseModel()
except TypeError as e:
    print(f"Error: {e}")
````

````python
# --- First Concrete Implementation ---
class DummyLinearRegression(BaseModel):
    """
    A dummy linear regression model that follows the BaseModel contract.
    """
    def train(self, X_train: pd.DataFrame, y_train: pd.Series):
        print("Training the linear regression model...")
        # In a real scenario, you would use scikit-learn here, e.g.:
        # from sklearn.linear_model import LinearRegression
        # self.model = LinearRegression()
        # self.model.fit(X_train, y_train)
        print("Linear regression model trained successfully!")

    def predict(self, X_test: pd.DataFrame) -> np.ndarray:
        print("Making predictions with the linear regression model...")
        # In a real scenario, you would call self.model.predict(X_test)
        # For this example, we'll just return a dummy array of zeros.
        return np.zeros(len(X_test))


# --- Second Concrete Implementation ---
class DummyRandomForest(BaseModel):
    """
    A dummy random forest model that also follows the BaseModel contract.
    """
    def train(self, X_train: pd.DataFrame, y_train: pd.Series):
        print("Training the random forest model...")
        # In a real scenario, you would use scikit-learn here, e.g.:
        # from sklearn.ensemble import RandomForestRegressor
        # self.model = RandomForestRegressor()
        # self.model.fit(X_train, y_train)
        print("Random forest model trained successfully!")

    def predict(self, X_test: pd.DataFrame) -> np.ndarray:
        print("Making predictions with the random forest model...")
        # For this example, we'll return a dummy array of ones.
        return np.ones(len(X_test))
````

````python
# The power of the base class is that you can now treat all your models in the same way, even though their internal logic is different. This is known as polymorphism.
# Create instances of our concrete models
lr_model = DummyLinearRegression()
rf_model = DummyRandomForest()

# Create some dummy data for the example
X_data = pd.DataFrame({'feature1': [1, 2, 3, 4, 5]})
y_data = pd.Series([10, 20, 30, 40, 50])

print("\n" + "="*40 + "\n")

# A list containing different types of models
models_to_run = [lr_model, rf_model]

# Loop through and use each model in exactly the same way
for model in models_to_run:
    print(f"--- Running workflow for: {model.get_info()} ---")
    
    # 1. Train the model
    model.train(X_data, y_data)
    
    # 2. Make predictions
    predictions = model.predict(X_data)
    
    print(f"Predictions: {predictions}")
    print("-" * 20 + "\n")
````

### 4.1 Decorators for Reusable Logic

A decorator is a function that takes another function as an argument, adds some kind of functionality, and then returns another function, all without altering the source code of the original function.

This is incredibly useful for cross-cutting concerns in a data pipeline. Instead of adding timing or logging code to every single function, you can write it once in a decorator and apply it where needed with a single line: `@time_execution`.

````python
import time
import numpy as np

def time_execution(func):
    """A decorator that prints the execution time of the decorated function."""
    def wrapper(*args, **kwargs):
        start_time = time.time()
        result = func(*args, **kwargs)
        end_time = time.time()
        print(f"'{func.__name__}' executed in {end_time - start_time:.4f} seconds.")
        return result
    return wrapper

# Let's create a sample data cleaning function
@time_execution
def clean_data(df: pd.DataFrame) -> pd.DataFrame:
    """A sample function that simulates a data cleaning operation."""
    # Simulate a time-consuming operation
    time.sleep(1)
    df['new_col'] = df['feature1'] * np.random.rand(len(df))
    time.sleep(1)
    return df

# Now, when we call clean_data, the decorator will automatically time it
print("--- Testing @time_execution decorator ---")
cleaned_df = clean_data(dummy_df)
````

A context manager is an object that defines a temporary context for a block of code. The `with` statement ensures that certain setup and teardown operations are performed, even if an error occurs within the block.

The most common use case is file handling. `with open(...)` guarantees that your file will be closed automatically, preventing data corruption or resource leaks, even if your script fails while writing results. This same pattern is used for managing database connections, ensuring the connection is closed properly.

### 4.3 A Glimpse at Design Patterns: Strategy Pattern

Design patterns are reusable, high-level blueprints for solving commonly occurring problems in software design. They are not specific algorithms, but rather general concepts.

The **Strategy Pattern** allows you to define a family of algorithms, put each of them into a separate class, and make their objects interchangeable. 

This is perfect for making a data processing pipeline flexible. Imagine you want to test different ways of imputing missing values (filling in NaNs). Instead of writing a giant `if/elif/else` block, you can use the Strategy pattern. You can define a `MeanImputer`, a `MedianImputer`, and a `KNNImputer`, and then configure your `DataProcessor` to use any of them interchangeably. This makes your code cleaner, easier to test, and much easier to extend with new imputation methods.

````python
from abc import ABC, abstractmethod
import numpy as np

# Create a DataFrame with missing values
nan_data = {'age': [25, 30, np.nan, 45], 'salary': [50000, np.nan, 70000, 80000]}
impute_df = pd.DataFrame(nan_data)

# 1. Define the abstract Strategy interface
class ImputationStrategy(ABC):
    @abstractmethod
    def impute(self, df: pd.DataFrame, column: str) -> pd.DataFrame:
        pass

# 2. Create Concrete Strategies
class MeanImputer(ImputationStrategy):
    def impute(self, df: pd.DataFrame, column: str) -> pd.DataFrame:
        print(f"-- Imputing '{column}' with Mean --")
        mean_val = df[column].mean()
        df[column] = df[column].fillna(mean_val)
        return df

class MedianImputer(ImputationStrategy):
    def impute(self, df: pd.DataFrame, column: str) -> pd.DataFrame:
        print(f"-- Imputing '{column}' with Median --")
        median_val = df[column].median()
        df[column] = df[column].fillna(median_val)
        return df

# 3. Create the Context class that uses a strategy
class DataProcessor:
    def __init__(self, strategy: ImputationStrategy):
        self._strategy = strategy
    
    def process_column(self, df: pd.DataFrame, column: str) -> pd.DataFrame:
        return self._strategy.impute(df, column)

# --- Demonstrate the pattern ---
print("Original DataFrame:")
print(impute_df)
print("\n")

# Use the MeanImputer strategy
mean_processor = DataProcessor(strategy=MeanImputer())
processed_df_mean = mean_processor.process_column(impute_df.copy(), 'salary')
print(processed_df_mean)
print("\n")

# Now, easily swap to the MedianImputer strategy for a different column
median_processor = DataProcessor(strategy=MedianImputer())
processed_df_median = median_processor.process_column(impute_df.copy(), 'age')
print(processed_df_median)
````

````python
from collections import Counter

# Imagine this list comes from a categorical column in a dataset
data_labels = ['cat', 'dog', 'cat', 'mouse', 'dog', 'cat', 'dog', 'cat']

# Use Counter to get the frequency of each label
label_counts = Counter(data_labels)

print(f"Label Frequencies: {label_counts}")

# You can easily find the most common items
print(f"The 2 most common labels are: {label_counts.most_common(2)}")
````
